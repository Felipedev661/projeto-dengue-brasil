```python

import zipfile
import pandas as pd

arquivo_zip = "/content/DENGBR23.csv.zip"
colunas = ["DT_SIN_PRI", "CLASSI_FIN", "CRITERIO"]

contagem_mensal = {}

with zipfile.ZipFile(arquivo_zip, "r") as z:
    with z.open("DENGBR23.csv") as f:
        for bloco in pd.read_csv(
            f,
            sep=",",
            encoding="latin1",
            usecols=colunas,
            dtype=str,
            chunksize=100_000
        ):
            filtro = (
                bloco["CLASSI_FIN"].isin(["10", "11", "12"])
                & bloco["CRITERIO"].isin(["1", "2"])
            )

            dados = bloco.loc[filtro, "DT_SIN_PRI"]
            datas = pd.to_datetime(dados, errors="coerce")

            meses = datas.dt.to_period("M").dropna().astype(str)
            contagem_mensal.update(
                meses.value_counts().to_dict()
            )

serie_mensal = (
    pd.Series(contagem_mensal, dtype="int64")
    .groupby(level=0).sum()
    .sort_index()
)

print("Registros por mês de início dos sintomas — 2023:")
display(serie_mensal.rename("Quantidade de registros").to_frame())
```

```python

arquivo_zip = "/content/DENGBR24.csv.zip"
contagem_mensal_2024 = {}

with zipfile.ZipFile(arquivo_zip, "r") as z:
    with z.open("DENGBR24.csv") as f:
        for bloco in pd.read_csv(
            f,
            sep=",",
            encoding="latin1",
            usecols=["DT_SIN_PRI", "CLASSI_FIN", "CRITERIO"],
            dtype=str,
            chunksize=100_000
        ):
            filtro = (
                bloco["CLASSI_FIN"].isin(["10", "11", "12"])
                & bloco["CRITERIO"].isin(["1", "2"])
            )

            datas = pd.to_datetime(
                bloco.loc[filtro, "DT_SIN_PRI"],
                errors="coerce"
            )

            meses = datas.dt.to_period("M").dropna().astype(str)
            frequencias = meses.value_counts()

            for mes, quantidade in frequencias.items():
                contagem_mensal_2024[mes] = (
                    contagem_mensal_2024.get(mes, 0) + quantidade
                )

serie_mensal_2024 = pd.Series(
    contagem_mensal_2024, dtype="int64"
).sort_index()

display(
    serie_mensal_2024.rename("Quantidade de registros").to_frame()
)
```

```python

for ano, caminho in arquivos.items():
    contagem = {}

    with zipfile.ZipFile(caminho, "r") as z:
        nome_csv = z.namelist()[0]

        with z.open(nome_csv) as f:
            for bloco in pd.read_csv(
                f,
                sep=",",
                encoding="latin1",
                usecols=["NU_ANO", "DT_SIN_PRI", "CLASSI_FIN", "CRITERIO"],
                dtype=str,
                chunksize=100_000
            ):
                filtro = (
                    bloco["CLASSI_FIN"].isin(["10", "11", "12"])
                    & bloco["CRITERIO"].isin(["1", "2"])
                )

                dados = bloco.loc[filtro, ["NU_ANO", "DT_SIN_PRI"]].copy()
                dados["ANO_SINTOMAS"] = pd.to_datetime(
                    dados["DT_SIN_PRI"], errors="coerce"
                ).dt.year.astype("Int64")

                resumo = dados.groupby(
                    ["NU_ANO", "ANO_SINTOMAS"], dropna=False
                ).size()

                for chave, quantidade in resumo.items():
                    contagem[chave] = contagem.get(chave, 0) + int(quantidade)

    print(f"\nArquivo {ano}:")
    tabela = pd.Series(contagem).rename("Registros").rename_axis(
        ["Ano de notificação", "Ano dos sintomas"]
    )
    display(tabela.to_frame())
```

```python

import zipfile
import pandas as pd
from collections import Counter

arquivos = {
    2023: "/content/DENGBR23.csv.zip",
    2024: "/content/DENGBR24.csv.zip",
    2025: "/content/DENGBR25.csv.zip"
}

contagens = Counter()

for ano_arquivo, caminho in arquivos.items():
    with zipfile.ZipFile(caminho) as z:
        with z.open(z.namelist()[0]) as f:
            for bloco in pd.read_csv(
                f,
                sep=",",
                encoding="latin1",
                usecols=["DT_SIN_PRI", "CLASSI_FIN", "CRITERIO"],
                dtype=str,
                chunksize=100_000
            ):
                filtro = (
                    bloco["CLASSI_FIN"].isin(["10", "11", "12"])
                    & bloco["CRITERIO"].isin(["1", "2"])
                )
                dados = bloco.loc[filtro].copy()
                dados["Data"] = pd.to_datetime(
                    dados["DT_SIN_PRI"], errors="coerce"
                )
                dados = dados.dropna(subset=["Data"])
                dados["Ano_sintomas"] = dados["Data"].dt.year
                dados["Mes"] = dados["Data"].dt.month

                agrupado = dados.groupby(
                    ["Ano_sintomas", "Mes", "CLASSI_FIN"]
                ).size()

                for chave, quantidade in agrupado.items():
                    contagens[chave] += int(quantidade)

base_resumida = pd.DataFrame(
    [
        {
            "Ano": chave[0],
            "Mes": chave[1],
            "Classificacao": chave[2],
            "Quantidade_registros": quantidade
        }
        for chave, quantidade in contagens.items()
    ]
).sort_values(["Ano", "Mes", "Classificacao"])

base_resumida.to_csv(
    "/content/dengue_resumo_mensal.csv",
    index=False,
    encoding="utf-8-sig"
)

display(base_resumida.head(10))
print("\nArquivo criado: dengue_resumo_mensal.csv")
print("Linhas na base resumida:", len(base_resumida))
```

```python
import zipfile
import pandas as pd

arquivos = {
    2023: "/content/DENGBR23.csv.zip",
    2024: "/content/DENGBR24.csv.zip",
    2025: "/content/DENGBR25.csv.zip"
}

for ano, caminho in arquivos.items():
    total_filtrado = 0
    marcados_duplicidade = 0

    with zipfile.ZipFile(caminho) as z:
        with z.open(z.namelist()[0]) as f:
            for bloco in pd.read_csv(
                f,
                sep=",",
                encoding="latin1",
                usecols=["CLASSI_FIN", "CRITERIO", "NDUPLIC_N"],
                dtype=str,
                chunksize=100_000
            ):
                filtro = (
                    bloco["CLASSI_FIN"].isin(["10", "11", "12"])
                    & bloco["CRITERIO"].isin(["1", "2"])
                )

                dados = bloco.loc[filtro]
                total_filtrado += len(dados)

                marcados_duplicidade += (
                    dados["NDUPLIC_N"].fillna("").str.strip() == "2"
                ).sum()

    print(f"\nArquivo {ano}")
    print("Total filtrado:", total_filtrado)
    print("Marcados como duplicidade (valor 2):",
          marcados_duplicidade)
```
